# Clean data

```python
# Add decade column for each df
dfs = [df_1950s, df_1960s, df_1970s, df_1980s, df_1990s, df_2000s, df_2010s]
decades = [1950, 1960, 1970, 1980, 1990, 2000, 2010]

for df in dfs:
    df["decade"] = decades.pop(0)
```

```python
# Combine the dataset
combined_df = pd.concat(dfs, ignore_index=True)
print(combined_df.shape)
```

```python
combined_df
```

```python
# Don't need the number column, relic from combining
combined_df = combined_df.drop(columns=["Number"])
```

```python
# Check for any rows where title and artist are the same
combined_df[combined_df.duplicated(["title", "artist"])]
# There's 7
```

```python
# Sort by year and keep earliest
combined_df = combined_df.sort_values("year", ascending=True)
combined_df = combined_df.drop_duplicates(subset=["title", "artist"], keep="first")
combined_df = combined_df.reset_index(drop=True)
```

# EDA

Goal here is to understand the target variable, feature distributions, and relationsips between features and popularity before any modelling decisions are made. Each visualisation directly motivates later modeling choices

```python
# Consistent style across all plots
sns.set_theme(style="whitegrid", palette="muted")
FIGSIZE_SINGLE = (8, 4)
FIGSIZE_WIDE   = (12, 5)
FIGSIZE_GRID   = (14, 10)
```

```python
### Popularity (target) distribution

fig, axes = plt.subplots(1, 2, figsize=FIGSIZE_WIDE)

# Histogram
axes[0].hist(combined_df["pop"], bins=30, color="#1DB954", edgecolor="white")
axes[0].set_title("Popularity score distribution")
axes[0].set_xlabel("Popularity (pop)")
axes[0].set_ylabel("Count")

# Add summary stats as a text box — useful for your write-up
stats = combined_df["pop"].describe()
textstr = "\n".join([
    f"Mean:   {stats['mean']:.1f}",
    f"Median: {stats['50%']:.1f}",
    f"Std:    {stats['std']:.1f}",
    f"Min:    {stats['min']:.0f}",
    f"Max:    {stats['max']:.0f}",
])

axes[0].text(0.97, 0.95, textstr, transform=axes[0].transAxes,
             fontsize=9, verticalalignment="top", horizontalalignment="right",
             bbox=dict(boxstyle="round", facecolor="white", alpha=0.7))

# Q-Q plot to formally assess normality
from scipy import stats as scipy_stats
scipy_stats.probplot(combined_df["pop"], dist="norm", plot=axes[1])
axes[1].set_title("Q-Q plot of popularity")

plt.suptitle("Figure 1: Target variable - popularity score", fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
### Popularity by decade

fig, ax = plt.subplots(figsize=FIGSIZE_WIDE)

decade_order = sorted(combined_df["decade"].unique())
sns.boxplot(
    data=combined_df,
    x="decade", y="pop",
    order=decade_order,
    palette="Blues",
    hue="decade",
    legend=False,
    ax=ax
)

# Overlay the mean per decade for clarity
decade_means = combined_df.groupby("decade")["pop"].mean()
ax.plot(range(len(decade_order)), [decade_means[d] for d in decade_order],
        "o--", color="tomato", label="Decade mean", zorder=5)

ax.set_title("Figure 2: Popularity score by decade")
ax.set_xlabel("Decade")
ax.set_ylabel("Popularity (pop)")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
### Correlation heatmap

numeric_cols = ["pop", "bpm", "nrgy", "dnce", "dB", "live",
                "val", "dur", "acous", "spch", "decade"]

corr_matrix = combined_df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(10, 8))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))  # hide upper triangle

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True, fmt=".2f",
    cmap="RdYlGn", center=0,
    vmin=-1, vmax=1,
    linewidths=0.5,
    ax=ax
)
ax.set_title("Figure 3: Feature correlation matrix")
plt.tight_layout()
plt.show()
```

```python
### Top feature scatter plots

# Automatically pick the top 4 features by absolute correlation with pop
# (excluding pop itself and decade, which gets its own plot above)
top_features = (
    corr_matrix["pop"]
    .drop(["pop", "decade"])
    .abs()
    .sort_values(ascending=False)
    .head(4)
    .index.tolist()
)

print(f"Top 4 features by correlation with pop: {top_features}")

fig, axes = plt.subplots(2, 2, figsize=FIGSIZE_GRID)
axes = axes.flatten()

for i, col in enumerate(top_features):
    axes[i].scatter(
        combined_df[col], combined_df["pop"],
        alpha=0.15, s=8, color="#1DB954"
    )
    # Fit and overlay a trend line
    m, b = np.polyfit(combined_df[col], combined_df["pop"], 1)
    x_line = np.linspace(combined_df[col].min(), combined_df[col].max(), 100)
    axes[i].plot(x_line, m * x_line + b, color="tomato", linewidth=1.5,
                 label=f"r = {corr_matrix.loc[col,'pop']:.2f}")
    axes[i].set_xlabel(col)
    axes[i].set_ylabel("pop")
    axes[i].set_title(f"{col} vs popularity")
    axes[i].legend(fontsize=9)

plt.suptitle("Figure 4: Top features vs popularity (with linear trend)",
             fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
### Genre distribution

genre_counts = combined_df["top genre"].value_counts()

fig, axes = plt.subplots(1, 2, figsize=FIGSIZE_WIDE)

# Top 15 genres by count
genre_counts.head(15).plot(
    kind="barh", ax=axes[0], color="#1DB954", edgecolor="white"
)
axes[0].invert_yaxis()
axes[0].set_title("Top 15 genres by song count")
axes[0].set_xlabel("Count")

# Cumulative coverage — helps choose N
cumulative_pct = (genre_counts.cumsum() / genre_counts.sum() * 100)
axes[1].plot(range(1, len(cumulative_pct) + 1), cumulative_pct.values,
             color="steelblue")
axes[1].axhline(80, color="tomato", linestyle="--", label="80% coverage")
axes[1].axhline(90, color="orange", linestyle="--", label="90% coverage")

# Find and annotate the N needed for 80% and 90%
n_80 = (cumulative_pct <= 80).sum() + 1
n_90 = (cumulative_pct <= 90).sum() + 1
axes[1].axvline(n_80, color="tomato",  linestyle=":", alpha=0.7)
axes[1].axvline(n_90, color="orange",  linestyle=":", alpha=0.7)
axes[1].set_xlabel("Number of genres (top N)")
axes[1].set_ylabel("Cumulative % of songs covered")
axes[1].set_title("Cumulative genre coverage")
axes[1].legend()

print(f"Top {n_80} genres cover 80% of songs")
print(f"Top {n_90} genres cover 90% of songs")

plt.suptitle("Figure 5: Genre distribution", fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
### EDA summary
# Print a concise summary to include verbatim in your write-up.

print("=" * 50)
print("EDA SUMMARY")
print("=" * 50)
print(f"Total songs:         {len(combined_df)}")
print(f"Popularity mean:     {combined_df['pop'].mean():.1f}")
print(f"Popularity median:   {combined_df['pop'].median():.1f}")
print(f"Decades covered:     {', '.join(str(d) for d in sorted(combined_df['decade'].unique()))}")
print(f"Unique genres:       {combined_df['top genre'].nunique()}")
print(f"Top feature (corr):  {top_features[0]}")
print(f"Genres for 80% cov:  {n_80}")
print(f"Genres for 90% cov:  {n_90}")
```

```python
### Genre vs Decade heatmap

# Get top 10 genres by count
top_genres = combined_df["top genre"].value_counts().head(10).index

# Filter to only those genres
df_genre_decade = combined_df[combined_df["top genre"].isin(top_genres)]

# Create a crosstab of genre vs decade (normalised by genre row)
crosstab = pd.crosstab(
    df_genre_decade["top genre"],
    df_genre_decade["decade"],
    normalize="index"  # each row sums to 1 — shows distribution across decades
) * 100  # as percentages

fig, ax = plt.subplots(figsize=(12, 6))
sns.heatmap(
    crosstab,
    annot=True, fmt=".0f",
    cmap="YlOrRd",
    linewidths=0.5,
    ax=ax
)
ax.set_title("Figure 6: Genre distribution across decades (% of each genre's songs)")
ax.set_xlabel("Decade")
ax.set_ylabel("Genre")
plt.tight_layout()
plt.show()
```

# Train Test Split

```python
# Train test and cv config
RANDOM_STATE  = 13   # unlucky for some
TEST_SIZE = 0.2
CV_FOLDS = 5
```

```python
# Feature selection and train test split

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

FEATURES = ["decade", "nrgy", "dnce", "dB", "dur", "acous", "spch"]
TARGET = "pop"

X = combined_df[FEATURES]
y = combined_df[TARGET]

print("Features shape: ", X.shape)
print("Target shape : ", y.shape)
```

```python
# Train test split of 80/20

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE
)

print("Training set: ", X_train.shape)
print("Test set: ", X_test.shape)
```

## Scale data for linear regression

```python
# scale for linear regression
# Fit ONLY on training data. Tree-based models will use X_train/X_test unscaled.

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)   # fit + transform on train
X_test_scaled  = scaler.transform(X_test)         # transform only on test

# Convert back to DataFrames to keep column names readable
X_train_scaled = pd.DataFrame(X_train_scaled, columns=FEATURES, index=X_train.index)
X_test_scaled  = pd.DataFrame(X_test_scaled,  columns=FEATURES, index=X_test.index)

print("Scaling applied. Verifying train set statistics:")
print(X_train_scaled.describe().round(2))
# Mean should be ~0 and std ~1 for all columns
```

# Modelling

```python
# Required Imports

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import cross_val_score
import numpy as np
```

```python
# Re-usable function for model performance metrics

def evaluate_model(name, y_train, y_pred_train, y_test, y_pred_test):
    metrics = {
        "Model": name,
        "Train RMSE": root_mean_squared_error(y_train, y_pred_train),
        "Test RMSE": root_mean_squared_error(y_test,  y_pred_test),
        "Train MAE": mean_absolute_error(y_train, y_pred_train),
        "Test MAE": mean_absolute_error(y_test,  y_pred_test),
        "Train R²": r2_score(y_train, y_pred_train),
        "Test R²": r2_score(y_test,  y_pred_test),
    }
    for k, v in metrics.items():
        print(f"  {k:12s}: {v:.4f}" if isinstance(v, float) else f"  {k:12s}: {v}")
    return metrics
```

## Linear Regression

```python
# Fit a basic linear regression model, no hp tuning
lr = LinearRegression()
lr.fit(X_train_scaled, y_train)
```

```python
# Store results of this and future models
results = []
results.append(evaluate_model(
    "Linear Regression",
    y_train,
    lr.predict(X_train_scaled),
    y_test,
    lr.predict(X_test_scaled)
))
```

```python
rmse_scores_lr = -cross_val_score(
    lr,
    X_train_scaled,
    y_train,
    cv=CV_FOLDS,
    scoring="neg_root_mean_squared_error"
)

print("=== LINEAR REGRESSION ===")
print("Fold RMSEs:", rmse_scores_lr)
print("Mean RMSE:", rmse_scores_lr.mean())
print("Std RMSE:", rmse_scores_lr.std())
```

```python
# Residuals plot

residuals_lr = y_test - lr.predict(X_test_scaled)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].scatter(lr.predict(X_test_scaled), residuals_lr, alpha=0.4, s=15, color="#1DB954")
axes[0].axhline(0, color="tomato", linestyle="--")
axes[0].set(title="Residuals vs predicted", xlabel="Predicted", ylabel="Residual")

axes[1].hist(residuals_lr, bins=25, color="#1DB954", edgecolor="white")
axes[1].axvline(0, color="tomato", linestyle="--")
axes[1].set(title="Residual distribution", xlabel="Residual", ylabel="Count")

plt.suptitle("Figure 7: Linear Regression — residual analysis", fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
# Coefficients - decide later whether to include
coef_df = pd.DataFrame({
    "feature":     FEATURES,
    "coefficient": lr.coef_
}).sort_values("coefficient", key=abs, ascending=False)

print(coef_df.to_string(index=False))
```

## Decision Tree

```python
# Update features set
FEATURES = ["decade", "nrgy", "dB", "dur", "acous", "spch"]

# Rebuild train/test sets with updated feature list
X_train = X_train[FEATURES]
X_test  = X_test[FEATURES]

# Scaled versions for any future linear models
X_train_scaled = X_train_scaled[FEATURES]
X_test_scaled  = X_test_scaled[FEATURES]
```

```python
# Fit a decision tree model, no hp tuning
dt = DecisionTreeRegressor(random_state=RANDOM_STATE)
dt.fit(X_train, y_train)
```

```python
results.append(evaluate_model(
    "Decision Tree",
    y_train, dt.predict(X_train),
    y_test,  dt.predict(X_test)
))
```

```python
rmse_scores_dt = -cross_val_score(
    lr,
    X_train_scaled,
    y_train,
    cv=CV_FOLDS,
    scoring="neg_root_mean_squared_error"
)

print("=== DECISION TREE  ===")
print("Fold RMSEs:", rmse_scores_dt)
print("Mean RMSE:", rmse_scores_dt.mean())
print("Std RMSE:", rmse_scores_dt.std())
```

```python
importance_df = pd.DataFrame({
    "feature":    FEATURES,
    "importance": dt.feature_importances_
}).sort_values("importance", ascending=False)

print(importance_df.to_string(index=False))
```

```python
fig, ax = plt.subplots(figsize=(20, 6))
plot_tree(
    dt,
    max_depth=3,
    feature_names=FEATURES,
    filled=True,
    rounded=True,
    fontsize=9,
    ax=ax
)
plt.suptitle("Figure 8: Decision Tree — first 3 levels", fontweight="bold")
plt.tight_layout()
plt.show()
```
